# GRR experiments index

This directory consolidates the experiments by manuscript placement and purpose.  Each analysis estimates both **ATE** and **ATT** whenever the implemented method supports both targets.  If a method is intrinsically target-specific, the executable notebook keeps the available target and records failures explicitly.

## Notebook organization

| Notebook | Manuscript placement | Purpose |
|---|---|---|
| `01_main_simulation_study.ipynb` | Main text simulation study | Three DGPs; compatible loss--link pairs; incompatible pairs; regularization paths. ATE and ATT are both reported. |
| `02_main_empirical_ihdp.ipynb` | Main text empirical study | IHDP semi-synthetic benchmark for ATE and ATT. |
| `03_main_empirical_lalonde_att.ipynb` | Main text empirical study | Lalonde/NSW analysis. ATT for NSW treated is the primary estimand, and ATE is reported as sensitivity. |
| `04_appendix_simulation_crossfit.ipynb` | Appendix simulation | Cross-fitting versus no cross-fitting for ATE and ATT. |
| `05_appendix_dimension_variation.ipynb` | Appendix simulation | Dimension variation for ATE and ATT. |
| `06_appendix_model_variation.ipynb` | Appendix simulation | Model and basis variation. GRR models report ATE and ATT; nearest-neighbor matching is kept as an ATE-only baseline when ATT is unsupported. |
| `07_appendix_additional_experiments.ipynb` | Appendix | Zhao/Kang--Schafer path, kernel-GP mismatch, ACIC, HDMA, and NSW checks, with ATE and ATT wherever possible. |
