# Generalized Riesz Regression experiments

This directory contains the publication-scale experiment notebooks for the GRR manuscript. The old exploratory notebooks, fast-mode code, synthetic data fallbacks, and external plotting helpers have been removed. Each notebook contains its own table and figure code so titles, labels, colors, line styles, and font sizes can be edited directly inside the notebook.

ATE and ATT are reported as separate estimands. Tables and figures are split by estimand rather than placing ATE and ATT side by side on the same axis or in the same summary table.

## Notebook map

| Notebook | Manuscript placement | Purpose |
|---|---|---|
| `01_main_simulation_study.ipynb` | Main text | Three-DGP simulation. Compatible loss-link comparison, incompatible loss-link comparison, and regularization paths. |
| `02_main_empirical_ihdp.ipynb` | Main text | IHDP semi-synthetic benchmark, estimating ATE and ATT in separate output blocks. |
| `03_main_empirical_lalonde.ipynb` | Main text | Lalonde job-training study, with ATT for NSW treated as the primary estimand and ATE as sensitivity. |
| `04_appendix_crossfit_comparison.ipynb` | Appendix | Compatible loss-link comparison with and without cross fitting. |
| `05_appendix_dimension_variation.ipynb` | Appendix | Dimension variation under compatible loss-link pairs. |
| `06_appendix_model_variation.ipynb` | Appendix | Basis/model variation: RKHS, polynomial, random forest leaves, random Fourier features, and nearest-neighbor matching where supported. |
| `07_appendix_score_guided_balancing.ipynb` | Appendix | Regressor balancing versus covariate balancing under score-guided balance. |
| `08_appendix_zhao_kernel_experiments.ipynb` | Appendix | Zhao/Kang--Schafer balance path and kernel-basis mismatch experiments. |
